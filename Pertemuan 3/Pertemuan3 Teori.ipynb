```python
# Menghitung rata-rata dari 3 bilangan bulat sembarang.
a = 10
b = 20 
c = 50

rata_rata = (a + b + c) / 3

print("Nilai Rata-rata dari ketiga bilangan adalaha",rata_rata)
```

```python
#Menghitung nilai fungsi f(x) = 2x3 + 2x + 15/x jika x merupakan bilangan bulat sembarang.
x = 30

hasil = 2 * (x ** 3) + 2 * x + (15 / x) 

print("Hasil dari fungsi f(x) = 2x3 + 2x + 15/x adalah",hasil)
```

```python
# Melakukan tukar nilai A,B,C,D menjadi B,D,A,C jika A,B,C,D merupakan bilangan desimal sembarang
A = 2.5
B = 3.5
C = 5.5
D = 4.5

print("Kondisi sebelum ditukar =",A,",",B,",",C,",",D)

A,B,C,D = B,D,A,C 

print("Kondisi Sesudah ditukar =",A,",",B,",",C,",",D)
```

```python
# membandingkan dua bilangan
x = 5
y = 10
print("Bilangan pertama x = %d \nbilangan kedua y = %d" %(x,y))
if(x > y) :
    print("X =",x)
    print("Bilangan terbesar")
if(x < y) :
    print("Y =",y)
    print("Bilangan terbesar")
if(x == y) :
    print("Bilangan sama yaitu :",x)
```

```python
umur = int(input("Masukkan Umur Anda :"))
if(umur <= 5) :
    kriteria = "Balita"
if(umur > 5 and umur <= 13) :
    kriteria = "Anak anak"
if(umur > 13 and umur <= 25) :
    kriteria = "Remaja"
if(umur > 25 and umur <= 35)  :
    kriteria = "Dewasa"
if(umur > 35 and umur <= 55) :
    kriteria = "Orang Tua"
if(umur > 55) :
    kriteria = "Lansia"
print("Umur = %d dan kriteria %a" %(umur,kriteria))
```
